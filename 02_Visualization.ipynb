```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Load cleaned dataset
df = pd.read_csv('data/fake_news_cleaned.csv')
print("Dataset loaded!", df.shape)
```

```python
# Plot fake vs real news count
plt.figure(figsize=(8, 5))
sns.countplot(x='label', data=df)
plt.title('Fake vs Real News Count')
plt.xlabel('Label (0 = Fake, 1 = Real)')
plt.ylabel('Count')
plt.show()
```

```python
# Plot subject distribution for fake news
plt.figure(figsize=(10, 5))
fake_subjects = df[df['label']==0]['subject'].value_counts()
sns.barplot(x=fake_subjects.index, y=fake_subjects.values)
plt.title('Fake News - Subject Distribution')
plt.xlabel('Subject')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Plot subject distribution for real news
plt.figure(figsize=(8, 5))
true_subjects = df[df['label']==1]['subject'].value_counts()
sns.barplot(x=true_subjects.index, y=true_subjects.values)
plt.title('Real News - Subject Distribution')
plt.xlabel('Subject')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Plot text length distribution for fake vs real news
plt.figure(figsize=(10, 5))
df['text_length'] = df['text'].apply(len)
sns.histplot(data=df, x='text_length', hue='label', bins=50)
plt.title('Text Length Distribution - Fake vs Real News')
plt.xlabel('Text Length')
plt.ylabel('Count')
plt.legend(['Real', 'Fake'])
plt.tight_layout()
plt.show()
```

```python
from wordcloud import WordCloud

# Word cloud for fake news
fake_text = ' '.join(df[df['label']==0]['text'].values)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(fake_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud - Fake News')
plt.show()
```

```python
# Word cloud for real news
real_text = ' '.join(df[df['label']==1]['text'].values)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(real_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud - Real News')
plt.show()
```
