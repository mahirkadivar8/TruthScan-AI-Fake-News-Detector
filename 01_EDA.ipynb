```python
# Import Libraries
import pandas as pd
import numpy as np  # used for future calculations
```

```python
fake = pd.read_csv('data/Fake.csv')
true = pd.read_csv('data/True.csv')

print("Fake news shape:", fake.shape)
print("True news shape:", true.shape)
```

```python
fake.head()
```

```python
true.head()
```

```python
# Add label 0 to fake news
fake['label'] = 0

# Add label 1 to true news
true['label'] = 1

print("Labels added!")
```

```python
# Combine both dataframes into one
df = pd.concat([fake, true], ignore_index=True)

print("Total rows:", len(df))
print("Columns:", df.columns.tolist())
```

```python
# Shuffle the data so fake and real news are mixed
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
print("Data shuffled!")
print(df.head())
```

```python
# Check basic info of dataset
df.info()
```

```python
# Check fake vs real news count
print("Fake vs Real News Count:")
print(df['label'].value_counts())
print("\n0 = Fake News")
print("1 = Real News")
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Check subject categories
print("Fake News Subjects:")
print(fake['subject'].value_counts())
print("\nTrue News Subjects:")
print(true['subject'].value_counts())
```

```python
# Save cleaned dataset
df.to_csv('data/fake_news_cleaned.csv', index=False)
print("Dataset saved successfully!")
```

```python
# Check average text length of fake vs real news
df['text_length'] = df['text'].apply(len)

print("Average text length:")
print(df.groupby('label')['text_length'].mean())
print("\n0 = Fake News, 1 = Real News")
```
